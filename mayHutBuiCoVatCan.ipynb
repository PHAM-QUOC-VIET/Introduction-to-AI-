```python
import random
```

```python
DIRECTIONS = {
    "UP":    (-1, 0),
    "DOWN":  (1, 0),
    "LEFT":  (0, -1),
    "RIGHT": (0, 1)
}
```

```python
class MatrixGenerator:
    def __init__(self, num_rows, num_cols, obstacle=0.15, dirt=0.35):
        self.num_rows = num_rows
        self.num_cols = num_cols
        self.obstacle = obstacle
        self.dirt = dirt

    def generate(self):
        matrix = []

        for i in range(self.num_rows):
            row = []

            for j in range(self.num_cols):
                # tạo ngẫu nhiên số thực từ [0, 1)
                rand = random.random() 

                # -1: vật cản, 1: có bụi, 0: sạch
                if rand < self.obstacle:
                    row.append(-1)
                elif rand < self.obstacle + self.dirt:
                    row.append(1)
                else:
                    row.append(0)

            matrix.append(row)

        return matrix
```

```python
class CleanerAgent:
    def __init__(self, matrix, start_x=0, start_y=0):
        self.matrix = matrix
        self.rows = len(matrix)
        self.cols = len(matrix[0])
        self.x = start_x
        self.y = start_y
        self.steps = 0

    # ktra vị trí hợp lệ
    def is_valid_position(self, x, y):
        return 0 <= x < self.rows and 0 <= y < self.cols

    # ktra có vật cản
    def is_obstacle(self, x, y):
        return self.matrix[x][y] == -1

    # các hướng đi có thể
    def possible_moves(self):

        moves = []

        for dir, (dx, dy) in DIRECTIONS.items():
            new_x, new_y = self.x + dx, self.y + dy

            # nếu vị trí hợp lệ và ko có vật cản và không phải là ô sạch (0)
            # thì có thể đi hướng đó
            if self.is_valid_position(new_x, new_y) and not self.is_obstacle(new_x, new_y):
                moves.append(dir)

        return moves

    # chọn hướng đi từ các hướng có thể đi
    def reflex_agent(self):
        moves = self.possible_moves()

        if not moves:
            return None
        
        return random.choice(moves)

    def act(self, action):
        if action is None:
            print('No valid moves available!!!')
            return
        
        # hút bụi trước khi di chuyển -> cập nhật matrix
        if self.matrix[self.x][self.y] == 1:
            self.matrix[self.x][self.y] = 0
            print(f'Cleaned at ({self.x}, {self.y})')

        # di chuyển
        # action thuộc (UP, DOWN, LEFT, RIGHT)
        dx, dy = DIRECTIONS[action]
        self.x += dx
        self.y += dy
        self.steps += 1

        print(f'Moved {action} -> ({self.x}, {self.y})')



    def is_clean(self):
        for row in self.matrix:
            if 1 in row:
                return False
        return True

    def run(self, max_steps=100):
        print("Matrix:")
        self.print_matrix()

        while not self.is_clean() and self.steps < max_steps:
            action = self.reflex_agent()
            self.act(action)
            self.print_matrix()

        print("Finished cleaning!")

    def print_matrix(self):
        for i in range(self.rows):
            
            for j in range(self.cols):

                if i == self.x and j == self.y:
                    print("X", end=" ")

                elif self.matrix[i][j] == -1:
                    print("#", end=" ")

                else:
                    print(self.matrix[i][j], end=" ")

            print()

        print()

    
```

```python
mat = MatrixGenerator(5, 5).generate()
agent = CleanerAgent(mat)
agent.run()
```
